```python
import os
import load_dotenv
from load_dotenv import load_dotenv

# This function will load all the variable from the .env file and 
# make them available in the os.environ dictionary (env variables)
load_dotenv()

if os.environ.get("CLAUDE_API_KEY"):
    print("API KEY variable has been loaded")
else:
    raise ValueError("CLAUDE API KEY not found")
```

#### Chain with Custom Runnable

```python
# TASK - 1 [Prompt]
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant"),
    ("human", "{input}")
])
```

```python
# TASK - 2 [LLM]
from langchain_anthropic import ChatAnthropic

llm_anthropic = ChatAnthropic(
    model=os.environ.get("CLAUDE_MODEL"), 
    temperature=0,
    api_key=os.environ.get("CLAUDE_API_KEY"),
    base_url=os.environ.get("CLAUDE_BASE_URL")
)
llm_anthropic
```

```python
# TASK - 3 [String Parser]
from langchain_core.output_parsers import StrOutputParser

str_parser = StrOutputParser()
```

```python
# TASK - 4 [Custom Runnable]
from langchain_core.runnables import RunnableLambda

def dictionary_maker(text:str)-> dict:

    return {"text" : text}

dictionary_maker_runnable = RunnableLambda(dictionary_maker)
```

```python
# TASK - 5 [Template For Post]

prompt_post = ChatPromptTemplate.from_messages(
    messages=[
        ("system", "You are a social media post generator"),
        ("human", "Create a post for the following text for LinkedIn: {text}")
    ]
)
```

```python
prompt_post.invoke({"text": "Hello World"})
```

```python
# TASK - 6 [Generate the Post]
from langchain_anthropic import ChatAnthropic

llm_anthropic = ChatAnthropic(
    model=os.environ.get("CLAUDE_MODEL"), 
    temperature=0,
    api_key=os.environ.get("CLAUDE_API_KEY"),
    base_url=os.environ.get("CLAUDE_BASE_URL")
)
llm_anthropic
```

```python
# TASK - 7 [Output Parser]
from langchain_core.output_parsers import StrOutputParser

str_parser = StrOutputParser()
```

```python
# CHAIN 

chain = prompt_template | llm_anthropic | str_parser | dictionary_maker_runnable | prompt_post | llm_anthropic | str_parser
chain.invoke("What is the capital of France ?")
```
