```python
import os
import load_dotenv
from load_dotenv import load_dotenv

# This function will load all the variable from the .env file and 
# make them available in the os.environ dictionary (env variables)
load_dotenv()

if os.environ.get("CLAUDE_API_KEY"):
    print("API KEY variable has been created")
else:
    raise ValueError("CLAUDE API KEY not found")
```

```python
from langchain_anthropic import ChatAnthropic

llm_anthropic = ChatAnthropic(
    model=os.environ.get("CLAUDE_MODEL"), 
    temperature=0,
    api_key=os.environ.get("CLAUDE_API_KEY"),
    base_url=os.environ.get("CLAUDE_BASE_URL")
)
llm_anthropic
```

#### **GUIDING IN PROMPTS**

```python
from langchain_core.prompts import PromptTemplate

result = llm_anthropic.invoke("Tell me a joke. Generate the output in key value pair format with the following keys: setup, punchline")
result.content
```

#### **Using Pydantic Models**

```python
from pydantic import BaseModel, Field # BaseModel Class creates the schema

class llm_schema(BaseModel):
    setup: str = Field(description = "The setup for the joke")
    punchline: str = Field(description = "The punchline for the joke")
```

```python
object = llm_schema(**{"setup": "some setup", "punchline": "some punchline"})
print(object)
print(object.setup)
```

```python
llm_structured_output = llm_anthropic.with_structured_output(llm_schema)

result = llm_structured_output.invoke("Tell me a joke")
print(result)

print(result.punchline)
```

```python
type(llm_schema) # This will return the pydantic type of the llm_schema class as shown below
# pydantic._internal._model_construction.ModelMetaclass
```

#### **Using TypedDic**

```python
from typing import TypedDict

class llm_schema_td(TypedDict):
    setup: str 
    punchline: str
```

```python
object = llm_schema_td(**{"ketchup": "some setup", "punchline": "some punchline"})
print(object)
# Even though I have given ketchup instead of setup it's not throwing error (Unlike Pydantic)
```

```python
llm_structured_output_td = llm_anthropic.with_structured_output(llm_schema_td)

result = llm_structured_output_td.invoke("Tell me a joke")
result # Returns the result in the form of dictionary but not in the form of pydantic object
```
